# Function to generate random characters/numbers
## Function to generate password

```python
import random
import string

def generate_password(length):
    lower = string.ascii_lowercase
    upper = string.ascii_uppercase
    numbers = string.digits
    special = string.punctuation

    all_characters = lower + upper + numbers + special

    password = "".join(random.choice(all_characters) for i in range(length))

    return password
```

# Read the password length from the User

```python
length = int(input("Enter the password length: "))

password = generate_password(length)

print("Generated Password:", password)
```

# Flask Application

```python
from flask import Flask, request

app = Flask(__name__)

@app.route('/')
def home():
    return '''
        <h2>Password Generator</h2>
        <form action="/generate">
            Enter Length: <input type="number" name="length">
            <input type="submit" value="Generate">
        </form>
    '''

@app.route('/generate')
def generate():
    length = int(request.args.get('length'))
    password = generate_password(length)
    return f"<h3>Your Password: {password}</h3>"

if __name__ == '__main__':
    app.run(debug=True, use_reloader=False)
```
